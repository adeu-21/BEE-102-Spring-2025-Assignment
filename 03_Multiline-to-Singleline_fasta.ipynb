Q3] Multi-line sequence fasta to single-line fasta file

```python
with open('multiline_input.fasta', 'r') as infile, open('singleline_output.fasta', 'w') as outfile:
    sequence = ''
    header = ''

    for line in infile:
        line = line.strip()

        if line.startswith('>'):
            if header: 
                outfile.write(header + '\n')
                outfile.write(sequence + '\n')
            header = line
            sequence = '' 
        else:
            sequence += line 

    if header:
        outfile.write(header + '\n')
        outfile.write(sequence + '\n')
```
