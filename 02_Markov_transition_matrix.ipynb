Q2] Building Markov transition matrix

```python
import pandas as pd
```

```python
# Input example sequence here
sequence = "ATGCGATCGATCGATCGATCG"
```

```python
nucleotides = ['A', 'C', 'G', 'T']
transition_counts = {a: {b: 0 for b in nucleotides} for a in nucleotides}

# Count pairs
for i in range(len(sequence) - 1):
    a, b = sequence[i], sequence[i + 1]
    if a in nucleotides and b in nucleotides:
        transition_counts[a][b] += 1
    else:
        print(f"Skip invalid: {a} or {b}")

# Normalize counts
prob = {a: {} for a in nucleotides}
for a in nucleotides:
    total = sum(transition_counts[a].values())
    for b in nucleotides:
        prob[a][b] = transition_counts[a][b] / total if total else 0.0


# Show transition matrix
df = pd.DataFrame(prob)
print("Transition Matrix:")
display(df)
```
